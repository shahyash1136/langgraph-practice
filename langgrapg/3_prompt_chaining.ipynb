```python
from langgraph.graph import StateGraph,START,END
from langchain_openai import ChatOpenAI
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
model = ChatOpenAI(model='gpt-4o-mini-2024-07-18',temperature=0)
```

```python
class BlogGenerateState(TypedDict):
    topic:str
    outline:str
    blog:str
    blog_score:str
```

```python
def generate_outline(state:BlogGenerateState)->BlogGenerateState:
    topic = state['topic']
    prompt = f"Generate outline line based on the topic = {topic}"
    outline = model.invoke(prompt).content
    state['outline'] = outline
    return state
```

```python
def generate_blog(state:BlogGenerateState)->BlogGenerateState:
    topic = state['topic']
    outline = state['outline']
    prompt = f"Generate detail blog based on the topic {topic} and outline {outline}"
    blog = model.invoke(prompt).content
    state['blog'] = blog
    return state
```

```python
def blog_evaluate(state:BlogGenerateState)->BlogGenerateState:
    outline = state['outline']
    blog = state['blog']

    prompt = f'Based on the {outline} rate my blog {blog} and give me the final score'

    score = model.invoke(prompt).content
    state['blog_score'] = score

    return state
```

```python
graph = StateGraph(BlogGenerateState)
graph.add_node('generate_outline',generate_outline)
graph.add_node('generate_blog',generate_blog)
graph.add_node('blog_evaluate',blog_evaluate)

graph.add_edge(START,'generate_outline')
graph.add_edge("generate_outline", "generate_blog")
graph.add_edge('generate_blog','blog_evaluate')
graph.add_edge("blog_evaluate", END)

workflow = graph.compile()
```

```python
initial_state = {'topic':'What is Langgraph?'}
final_state = workflow.invoke(initial_state)
print(final_state)
```
